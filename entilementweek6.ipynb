```python
# File: USN_Week6_PropositionalLogic_TruthTableEnumeration.py

import itertools

def evaluate(expr, model):
    """Evaluate a propositional expression under a given model."""
    return eval(expr, {}, model)

def entails(kb, query, symbols):
    """Check if KB entails query using truth-table enumeration."""
    for values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        kb_true = all(evaluate(expr, model) for expr in kb)
        if kb_true and not evaluate(query, model):
            return False
    return True

if __name__ == "__main__":
    # Define KB in propositional form
    kb = [
        "(P == Q)",     # P ↔ Q
        "(Q or R)",     # Q ∨ R
        "(P or R)"      # ¬P → R is equivalent to P ∨ R
    ]
    symbols = ["P", "Q", "R"]

    # Define queries
    queries = {
        "Q": "Q",
        "P->R": "(not P) or R",
        "R->Q": "(not R) or Q"
    }

    # Check entailment for each query
    for name, expr in queries.items():
        result = entails(kb, expr, symbols)
        print(f"Does KB entail {name}? {result}")
    print("1WN24CS134")
```
