```python
def alpha_beta(depth, node, alpha, beta, maximizing):

    if depth == 0:
        return node

    if maximizing:
        best = -999

        for child in node:
            value = alpha_beta(depth - 1, child,
                               alpha, beta, False)

            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                break

        return best

    else:
        best = 999

        for child in node:
            value = alpha_beta(depth - 1, child,
                               alpha, beta, True)

            best = min(best, value)
            beta = min(beta, best)

            if alpha >= beta:
                break

        return best


# Taking input from user
n = int(input(" 1WN24CS134 \Enter number of leaf nodes: "))

values = list(map(int, input("Enter leaf node values: ").split()))

tree = []

for i in range(0, n, 2):
    tree.append(values[i:i+2])

result = alpha_beta(2, tree, -999, 999, True)

print("Best value:", result)
```
