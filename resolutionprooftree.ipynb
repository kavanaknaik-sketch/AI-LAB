```python

def negate(literal):
    if literal.startswith("~"):
        return literal[1:]
    return "~" + literal


def resolution_proof_tree(kb):
    clauses = set(frozenset(c) for c in kb)
    proof_tree = {}

    while True:
        new_clauses = set()

        for c1 in clauses:
            for c2 in clauses:

                if c1 == c2:
                    continue

                for literal in c1:
                    opposite = negate(literal)

                    if opposite in c2:
                        r = (c1 - {literal}) | (c2 - {opposite})

                        # Avoid tautological clauses
                        if any(negate(x) in r for x in r):
                            continue

                        if r not in clauses:
                            new_clauses.add(r)
                            proof_tree[r] = (c1, c2)

                        if len(r) == 0:
                            print("Goal Proved! 1WN24CS134")
                            print("\nProof Tree:")

                            for result, parents in proof_tree.items():
                                print(parents, "=>", result)

                            return

        if not new_clauses:
            print("Goal Cannot Be Proved")
            return

        clauses.update(new_clauses)


# Example:
# If it rains, the ground becomes wet.
# It rains.
# Goal: Ground is wet.

kb = [
    {"~Rain", "Wet"},
    {"Rain"},
    {"~Wet"}
]

resolution_proof_tree(kb)
```
