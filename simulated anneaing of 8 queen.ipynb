```python
import math
import random

def get_heuristic(board):
    """Calculates the number of attacking queen pairs."""
    conflicts = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def print_board(board):
    """Prints a visual representation of the N-Queens board."""
    n = len(board)
    for row in range(n):
        line = ""
        for col in range(n):
            line += " Q " if board[col] == row else " . "
        print(line)

def simulated_annealing_8_queens():
    # Initial state: random row position for each column (0 to 7)
    current_board = [random.randint(0, 7) for _ in range(8)]
    current_h = get_heuristic(current_board)

    # Annealing parameters
    temperature = 1000.0
    cooling_rate = 0.995

    while current_h > 0:
        if temperature <= 0.0001:
            # Restart with a new initial state if cooled without finding a solution
            current_board = [random.randint(0, 7) for _ in range(8)]
            current_h = get_heuristic(current_board)
            temperature = 1000.0

        # Pick a random neighbor: choose a random column and a new random row
        col = random.randint(0, 7)
        new_row = random.randint(0, 7)

        while new_row == current_board[col]:
            new_row = random.randint(0, 7)

        next_board = list(current_board)
        next_board[col] = new_row
        next_h = get_heuristic(next_board)

        delta = next_h - current_h

        # Accept if better, or accept worse move with a probability based on Temperature
        if delta < 0 or random.random() < math.exp(-delta / temperature):
            current_board = next_board
            current_h = next_h

        temperature *= cooling_rate

    # Output ONLY the final solution
    print("==========================================")
    print("      8-QUEENS SIMULATED ANNEALING    1WN24CS134 Kavana K Naik    ")
    print("==========================================")
    print(f"Solution Board (cols 0-7): {current_board}")
    print(f"Attacking pairs (Heuristic h): {current_h}\n")
    print_board(current_board)

if __name__ == "__main__":
    simulated_annealing_8_queens()
```
