```python
import random

def get_heuristic(board):
    """
    Calculates the number of attacking queen pairs.
    board[i] represents the row position of the queen in column i.
    """
    conflicts = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row conflict
            if board[i] == board[j]:
                conflicts += 1
            # Diagonal conflict (row difference == column difference)
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def print_board(board):
    """Prints a visual representation of the N-Queens board."""
    n = len(board)
    for row in range(n):
        line = ""
        for col in range(n):
            if board[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
    print()

def get_best_neighbor(board):
    """
    Evaluates all possible single-queen row moves (neighbors)
    and returns the neighbor with the lowest heuristic score.
    """
    current_h = get_heuristic(board)
    best_board = list(board)
    best_h = current_h

    n = len(board)
    neighbors = []

    # Explore moving each queen to every other row in its column
    for col in range(n):
        original_row = board[col]
        for row in range(n):
            if row != original_row:
                neighbor = list(board)
                neighbor[col] = row
                h = get_heuristic(neighbor)
                neighbors.append((h, neighbor))

    # Find the minimum heuristic score among neighbors
    min_h = min(neighbors, key=lambda x: x[0])[0]

    # Collect all neighbors tied for the lowest score
    best_neighbors = [b for h, b in neighbors if h == min_h]

    # Choose one best neighbor (randomly pick if tied)
    best_board = random.choice(best_neighbors)
    best_h = min_h

    return best_board, best_h

def hill_climbing_4_queens():
    """Executes the standard steepest-ascent hill climbing search."""
    # Generate a random initial board configuration: board[col] = row
    current_board = [random.randint(0, 3) for _ in range(4)]
    current_h = get_heuristic(current_board)

    step = 0
    print("==========================================")
    print("     4-QUEENS HILL CLIMBING ALGORITHM  1WN24CS134 Kavana K Naik    ")
    print("==========================================")
    print(f"Step {step}: Initial State")
    print(f"Board representation (cols 0-3): {current_board}")
    print(f"Attacking pairs (Heuristic h): {current_h}\n")
    print_board(current_board)

    while True:
        if current_h == 0:
            print("==========================================")
            print(" SUCCESS: Found a valid solution!")
            print("==========================================")
            break

        neighbor, neighbor_h = get_best_neighbor(current_board)

        # If no neighbor improves the heuristic, we are stuck (Local Optimum or Shoulder)
        if neighbor_h >= current_h:
            print("==========================================")
            print(" STUCK: Reached a Local Optimum / Plateau.")
            print(" No neighbor has a lower heuristic value.")
            print("==========================================")
            break

        # Move to the best neighbor
        current_board = neighbor
        current_h = neighbor_h
        step += 1

        print(f"Step {step}: Moved to better state")
        print(f"Board representation (cols 0-3): {current_board}")
        print(f"Attacking pairs (Heuristic h): {current_h}\n")
        print_board(current_board)

if __name__ == "__main__":
    hill_climbing_4_queens()
```
