```python
import heapq

# Configuration set strictly for a 3-step solution path
INITIAL_STATE = (1, 2, 3, 0, 4, 6, 7, 5, 8)
GOAL_STATE    = (1, 2, 3, 4, 5, 6, 7, 8, 0)

class PuzzleNode:
    def __init__(self, state, g_cost, parent=None):
        self.state = state        # Current layout tuple
        self.g_cost = g_cost      # Node depth g(n)
        self.parent = parent      # Reference to reconstruct path steps
        self.h_cost = self._calculate_misplaced()
        self.f_cost = self.g_cost + self.h_cost  # f(n) = g(n) + h(n)

    def _calculate_misplaced(self):
        # Counts mismatched tiles from their goal slots (ignoring blank space '0')
        return sum(1 for i in range(9) if self.state[i] != 0 and self.state[i] != GOAL_STATE[i])

    def __lt__(self, other):
        return self.f_cost < other.f_cost

def get_neighbors(state):
    """Computes adjacent valid tiles moves by sliding the blank position (0)."""
    neighbors = []
    blank_idx = state.index(0)
    row, col = blank_idx // 3, blank_idx % 3

    # Grid offset mapping: Up, Down, Left, Right
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col
            new_state = list(state)
            new_state[blank_idx], new_state[new_idx] = new_state[new_idx], new_state[blank_idx]
            neighbors.append(tuple(new_state))

    return neighbors

def solve_3_steps():
    """Solves the pre-configured layout using A* search."""
    start_node = PuzzleNode(INITIAL_STATE, 0)
    open_list = [start_node]
    closed_set = set()

    while open_list:
        current_node = heapq.heappop(open_list)

        if current_node.state == GOAL_STATE:
            # Reconstruct the forward solution step array
            path = []
            while current_node:
                path.append(current_node.state)
                current_node = current_node.parent
            return path[::-1]

        closed_set.add(current_node.state)

        for neighbor_state in get_neighbors(current_node.state):
            if neighbor_state in closed_set:
                continue

            neighbor_node = PuzzleNode(neighbor_state, current_node.g_cost + 1, current_node)
            heapq.heappush(open_list, neighbor_node)

    return None

def print_board(state):
    """Prints the puzzle tuple formatted gracefully as a 3x3 matrix."""
    for i in range(0, 9, 3):
        print(f" {state[i]}  {state[i+1]}  {state[i+2]}")
    print()

if __name__ == "__main__":
    print("=== 8-Puzzle A* Framework: 3-Step Execution ===\n")

    solution_path = solve_3_steps()

    if solution_path:
        print(f"✅ Goal path tracked successfully!")
        print(f"📊 Total transitions: {len(solution_path) - 1} steps\n")

        for step, state in enumerate(solution_path):
            if step == 0:
                print(f"[Start: Initial State]")
            elif step == len(solution_path) - 1:
                print(f"[Step {step}: Goal State Reached!]")
            else:
                print(f"[Step {step}]")
            print_board(state)
    else:
        print("❌ Error generating path sequence.")
print("1WN24CS134 Kavana K Naik")
```
