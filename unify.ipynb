```python
def parse_term(term):
    term = term.strip()

    if "(" not in term:
        return term

    name, args = term.split("(", 1)
    args = args[:-1]

    arguments = []
    depth = 0
    current = ""

    for ch in args:
        if ch == "," and depth == 0:
            arguments.append(parse_term(current))
            current = ""
        else:
            current += ch
            if ch == "(":
                depth += 1
            elif ch == ")":
                depth -= 1

    if current.strip():
        arguments.append(parse_term(current))

    return (name.strip(), arguments)


def is_variable(x):
    return isinstance(x, str) and x[0].islower()


def apply_substitution(term, subst):
    while isinstance(term, str) and term in subst:
        term = subst[term]

    if isinstance(term, tuple):
        return (term[0], [
            apply_substitution(arg, subst)
            for arg in term[1]
        ])

    return term


def occurs_check(var, term, subst):
    term = apply_substitution(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(
            occurs_check(var, arg, subst)
            for arg in term[1]
        )

    return False


def unify(x, y, subst):
    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return True

    if is_variable(x):
        if occurs_check(x, y, subst):
            return False
        subst[x] = y
        return True

    if is_variable(y):
        if occurs_check(y, x, subst):
            return False
        subst[y] = x
        return True

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return False

        for a, b in zip(x[1], y[1]):
            if not unify(a, b, subst):
                return False

        return True

    return False


# Main Program
exp1 = input(" 1WN24CS134 Enter first expression: ")
exp2 = input("Enter second expression: ")

term1 = parse_term(exp1)
term2 = parse_term(exp2)

substitution = {}

if unify(term1, term2, substitution):
    print("\nUnification Successful!")
    print("Substitution Set:")

    if substitution:
        for key, value in substitution.items():
            print(key, "=", value)
    else:
        print("NIL")
else:
    print("\nUnification Failed!")
```
