```python

import random
import math

def aco_tsp(distances, num_ants=10, iterations=100,
            alpha=1.0, beta=2.0, rho=0.5, Q=100):

    n = len(distances)

    # 1. Initialize pheromone trails
    pheromone = [[1.0 for _ in range(n)] for _ in range(n)]

    best_tour = None
    best_length = float("inf")

    # 2. Repeat for a fixed number of iterations
    for iteration in range(iterations):
        all_tours = []
        all_lengths = []

        # 3. Each ant constructs a tour
        for ant in range(num_ants):
            start = random.randrange(n)
            tour = [start]
            unvisited = set(range(n))
            unvisited.remove(start)

            while unvisited:
                current = tour[-1]

                # Calculate probabilities for unvisited cities
                candidates = list(unvisited)
                weights = []

                for city in candidates:
                    tau = pheromone[current][city] ** alpha
                    eta = (1.0 / distances[current][city]) ** beta
                    weights.append(tau * eta)

                total = sum(weights)
                probabilities = [w / total for w in weights]

                # Select next city probabilistically
                next_city = random.choices(
                    candidates, weights=probabilities, k=1
                )[0]

                tour.append(next_city)
                unvisited.remove(next_city)

            # Return to the starting city
            tour.append(start)

            # 4. Calculate total tour length
            length = sum(
                distances[tour[i]][tour[i + 1]]
                for i in range(n)
            )

            all_tours.append(tour)
            all_lengths.append(length)

            # Record the best tour found so far
            if length < best_length:
                best_length = length
                best_tour = tour[:]

        # 5. Evaporate pheromone
        for i in range(n):
            for j in range(n):
                pheromone[i][j] *= (1 - rho)

        # 6. Deposit pheromone based on tour quality
        for tour, length in zip(all_tours, all_lengths):
            deposit = Q / length

            for i in range(n):
                a = tour[i]
                b = tour[i + 1]

                pheromone[a][b] += deposit
                pheromone[b][a] += deposit

    # 7. Return the best solution
    return best_tour, best_length


# Example: Distance matrix for 5 cities
distances = [
    [0,  10, 15, 20, 25],
    [10,  0, 35, 25, 17],
    [15, 35,  0, 30, 28],
    [20, 25, 30,  0, 12],
    [25, 17, 28, 12,  0]
]

# Run ACO
best_tour, best_length = aco_tsp(
    distances,
    num_ants=10,
    iterations=100,
    alpha=1.0,
    beta=2.0,
    rho=0.5,
    Q=100
)

print("Best Tour:", " -> ".join(map(str, best_tour)))
print("Shortest Tour Length:", best_length)
print("1WN24CS134")
```
